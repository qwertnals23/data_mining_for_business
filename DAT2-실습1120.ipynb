```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
sales = pd.read_csv('DepartmentStoreSales.csv')
sales
```

```python
plt.plot(sales.Quarter, sales.Sales)
```

```python
import statsmodels.formula.api as sm
from statsmodels.tsa import tsatools

sales_df = tsatools.add_trend(sales, trend='ctt')
# fit a linear regression model to the time series
sales_lm = sm.ols(formula='Sales ~ trend + trend_squared',
data=sales_df).fit()
plt.plot(sales.Quarter, sales.Sales)
sales_lm.predict(sales_df).plot()
```

```python
sales_df = tsatools.add_trend(sales, trend='ct')
# fit a linear regression model to the time series
sales_lm = sm.ols(formula='Sales ~ trend',
data=sales_df).fit()
plt.plot(sales.Quarter, sales.Sales)
sales_lm.predict(sales_df).plot()
```
