### 감성 사전 활용

```python
import nltk
import pandas as pd

nltk.download('opinion_lexicon')

from nltk.corpus import opinion_lexicon

pos_list=set(opinion_lexicon.positive())
neg_list=set(opinion_lexicon.negative())

pos_list
```

```python
neg_list
```

```python
from nltk.tokenize import RegexpTokenizer

tokenizer = RegexpTokenizer("[\w]+")

def sentiment(sentence):
  senti=0
  words = [word.lower() for word in tokenizer.tokenize(sentence)]
  for word in words:
    if word in pos_list:
      senti += 1
    elif word in neg_list:
      senti -= 1
  return senti

from tqdm import *

scores = []
x = len(mobile)

for i in tqdm(range(0,x)):
    text = mobile.iloc[i,4]
    if pd.isnull(text):
        scores.append(0)
    else :
        scores.append(sentiment(text))

mobile['sentiment_dict'] = scores
mobile.head()
```

```python
import pandas as pd
import numpy as np
from transformers import AutoTokenizer, AutoModelForSequenceClassification, pipeline

# 1) 모델 이름 설정 (영어 감성분석용 BERT)
model_name = "distilbert-base-uncased-finetuned-sst-2-english"

# 2) 토크나이저 & 모델 로드
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)

# 3) 파이프라인 생성
sentiment_pipeline = pipeline(
    "sentiment-analysis",
    model=model,
    tokenizer=tokenizer
)
```

```python
def bert_sentiment_score(text):
    # 결측값/비문자 처리
    if pd.isna(text):
        return 0.0
    if not isinstance(text, str):
        text = str(text)
    text = text.strip()
    if text == "":
        return 0.0
    
    # BERT는 최대 토큰 길이가 있으므로, 너무 긴 문장은 512자 정도로 자름
    text = text[:512]
    
    # 파이프라인 실행
    result = sentiment_pipeline(text)[0]
    label = result["label"]       # "POSITIVE" 또는 "NEGATIVE"
    score = result["score"]       # 0 ~ 1 사이 확률값

    # 부호 붙여서 하나의 수로 변환 
    if label == "NEGATIVE":
        score = -score
    return score
```

```python
sent_bert = []

for i in range(len(mobile)):
    text = mobile.iloc[i, 4]      
    sent_bert.append(bert_sentiment_score(text))

mobile['bert_sentiment_score'] = sent_bert
mobile.head()
```
